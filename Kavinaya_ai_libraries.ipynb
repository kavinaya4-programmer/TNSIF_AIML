```python
import numpy as np

marks = np.array([[67,55,80],[23,45,67],[90,88,92],[20,30,40],[50,60,70]])

print("\nStudent Marks")
print(marks)

print("\nAverage of Subject - Wise")

print(np.mean(marks, axis = 0))

print("\nAverage of Student - Wise")

print(np.mean(marks, axis = 1))



print("\nHighest Mark: ", np.max(marks))
print("\nLowest Mark: ", np.min(marks))

average = np.mean(marks, axis = 1)

for i in range(len(average)):
  if average[i] > 65:
    print("Student ",i)

```

```python
import pandas as pd

datas = {
    "Employee" : ["Abbas","Heman","Hari","Ammar","Shalini"],
    "Department":["IT","Marketing", "Finance","Marketing","IT"],
    "Salary":[85000,45000,75000,45000,55000]
}

df = pd.DataFrame(datas)

print(df)

print("\nIT Department")
print(df[df["Department"] == "IT"])

print("\nHighest Salary")
print(df[df["Salary"] == df["Salary"].max()])

print("\nAfter 10% increment")
df["Updated Salary"] = df["Salary"] * 1.10

print(df)

```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [120,150,170,160,210]

# Line Plot
plt.figure(figsize=(6,4))
plt.plot(months,sales,marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# Bar Plot
plt.figure(figsize=(6,4))
plt.bar(months,sales)
plt.title("Monthly Sales")
plt.show()

# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(months,sales)
plt.title("Sales Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

print(tips.head())

# Heatmap
correlation = tips.corr(numeric_only=True)

plt.figure(figsize=(6,4))
sns.heatmap(correlation,
            annot=True,
            cmap="coolwarm")

plt.show()

# Pair Plot
sns.pairplot(
    tips,
    hue="sex"
)

plt.show()
```

```python
import pandas as pd
import plotly.express as px

data = {
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[120,180,150,220,260],
    "Profit":[20,35,30,45,60]
}

df = pd.DataFrame(data)

fig = px.line(
    df,
    x="Month",
    y=["Sales","Profit"],
    title="Sales & Profit Dashboard",
    markers=True
)

fig.show()
```
